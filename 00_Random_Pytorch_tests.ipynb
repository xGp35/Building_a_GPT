### Vector Quantization:

Example to show real use of broadcasting <br>

The basic operation in `Vector Quantization` finds the closest point in a set of points, (called `codes` in VQ jargon), to a given point, called the `observation`.

```python
observation = np.array([111.0, 188.0])
```

```python
codes = np.array([[102.0, 203.0],
                  [132.0, 193.0],
                  [45.0, 155.0],
                  [57.0, 173.0]])
```

```python
diff = codes - observation  # The broadcast happens here.
print(diff)
```

```python
diff**2
```

```python
np.sum(diff**2, axis = 1)
```

```python
dist = np.sqrt(np.sum(diff**2, axis = 1))
print(dist)
```

```python
desired_index = np.argmin(dist)
print(desired_index)
```
